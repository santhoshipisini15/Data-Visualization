```python
import matplotlib.pyplot as plt

subjects = ['Python' , 'Java' , 'C++' , 'SQL']
students = [45, 30 ,38 , 50]

plt.bar(subjects, students)

plt.title("Students Enrolled")
plt.xlabel("Subjects")
plt.ylabel("Students")

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [10,20,30,40,50]

plt.plot(x,y , marker='^')

plt.title("triangle  Maker")

plt.show()
```

```python
import matplotlib.pyplot as plt

study_hours = [1,2,3,4,5,6,7,8]
marks = [60,70,80,75,85,90,95,100]

plt.scatter(study_hours,marks)

plt.title("Study Hours vs Marks")

plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
import matplotlib.pyplot as plt

months = ['Jan','Feb','Mar','Apr','May']
sales = [25,30,40,35,50]

plt.plot(months, sales)

plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.show()
```

```python
from matplotlib.lines import lineStyles

plt.plot(months, sales ,
         marker='o' ,
         linestyle='--' ,
         color='blue')

plt.show()
```

```python
import matplotlib.pyplot as plt

age = [20,25,30,35,40,45,50]
salary = [25000,30000,35000,40000,45000,50000,55000]

plt.scatter(age, salary)
plt.figure(figsize=(10,6))

plt.title("Age vs Salary")

plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()
```
